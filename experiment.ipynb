```python
bin_width=5
bins=np.arange(df["Age"].min(),df["Age"].max())
fig, ax= plt.subplots(dpi=300)
sns.histplot(
    data=df,
    x="Age",
    hue="Survived",
    multiple="stack",
    bins=bins,
    ax=ax
)

age_bins = pd.cut(df["Age"], bins=bins, include_lowest=True)
grouped = pd.crosstab(age_bins, df["Survived"])
ratio = grouped[1]
```
