### Phase 1: ベースライン構築 & 前処理* **欠損値処理**: * `Age`, `Fare`: 中央値（median）で補完 * `Embarked`: 最頻値（mode）で補完* **外れ値処理**: * `Age` の上位1%（99パーセンタイル）をキャッピング処理* **エンコーディング**: * `Sex_Pclass`（性別×客室クラス）の結合カテゴリを作成し、One-Hot Encoding を適用 * `Sex` は OrdinalEncoder にて数値化### Phase 2: 特徴量エンジニアリング（突破口）* **`boy_pclass2` フラグの追加**:

* 条件: `Age <= 10` 且つ `Sex == male` 且つ `Pclass == 2`

* 考察: 2等客室の男の子の生存率傾向に着目した特徴量を作成し、モデルの予測精度が 0.77 に向上。
